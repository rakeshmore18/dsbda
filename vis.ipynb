```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn
```

```python
df=pd.read_csv('forestfires.csv')
df
```

1. Using Matplotlib

```python
#Scatter Plot

plt.scatter(df['month'], df['wind'])
 
# Adding Title to the Plot
plt.title("Scatter Plot")
 
# Setting the X and Y labels
plt.xlabel('Month')
plt.ylabel('wind')

 
plt.show()
```

```python
#Bar Chart

#Bar chart with day against temprature
plt.bar(df['day'], df['temp'])
 
plt.title("Bar Chart")
 
# Setting the X and Y labels
plt.xlabel('Day')
plt.ylabel('Temprature')
 
# Adding the legends
plt.show()
```

```python
#Histogram

# histogram of total_bills
plt.hist(df['DMC'])
 
plt.title("Histogram")
 
# Adding the legends
plt.show()
```

```python
#Pie Chart

keys='FFMC','DMC','DC'
sizes=[df['FFMC'].sum(),df['DMC'].sum(),df['DC'].sum()]
# Plotting data on chart



plt.pie(sizes, labels=keys, autopct='%.0f%%')

# Add title to the chart
plt.title('Values of fire weather indices')

# Displaying chart
plt.show() 
```

```python
#Box Plot
df
df.boxplot(by ='day', column =['temp'], grid = False)
```

2. Using Seaborn Library

```python
df
```

```python
#Scatter Chart

sn.scatterplot(x='month', y='temp', data=df,hue='day')
plt.title('Scatter Chart')
plt.show()
```

```python
# Line Plot

# draw lineplot
sn.lineplot(x="month", y="temp", data=df)
 
# setting the title 
plt.title('Line Chart')
 
plt.show()
```

```python
# Draw a vertical boxplot grouped 
# by a categorical variable:
sn.set_style("whitegrid")
  
sn.boxplot(x = 'day', y = 'temp', data = df)
```
