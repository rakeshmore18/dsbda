```python

#handling outliers

def remove_outliers(df,coloumns,n_std):
  for col in coloumns:
      print('Working on coloumn: {}'.format(col))
        
      mean = df[col].mean()
      sd = df[col].std()
        
      df = df[(df[col] <= mean+(n_std*sd))]
  return df

df

```
