```python
#import the libraries
import pandas as pd
import numpy as np
```

```python
#Read dataset
df=pd.read_csv('dataset_Facebook.csv',sep=';')
```

```python
df
```

```python
#a. Create data subsets

#1.	Create a subset of a Python dataframe using the loc() function
#Extract data of specific rows of a dataframe
df.loc[[1,3,5]]
```

```python
#Create a subset of rows using slicing
df.loc[0:2:10]
```

```python
#Create a subset of particular columns using labels
df.loc[0:5,['Type','Category']]
```

```python
#2. Using Python iloc() function to create a subset of a dataframe
df.iloc[0:5,[0,2]]
```

```python
#3. Indexing operator to create a subset of a dataframe
df[['Page total likes','Type','Category','comment']]
```

```python
#b. Merge Data

df1=df[['Page total likes','Type','Category','comment']]
df1
```

```python
df2=df[['Post Month','Post Weekday','Post Hour','Paid']]
df2
```

```python
pd.concat([df1,df2])
```

```python
#c. Sort Data

#Sorting the Data frame in ascending order 

df.sort_values(by=['Category'])
```

```python
#Sorting the Data frame in descending order 

df.sort_values(by=['Category'], ascending=False)
```

```python
#Sorting Pandas Data frame by putting missing values first

df.sort_values(by=['Category'], na_position='first')
```

```python
#Sorting Data frames by multiple columns but different order

df.sort_values(by=['Type', 'Category'],ascending=[False, True])
```

```python
#d. Transposing Data
result = df.transpose()
result
```
