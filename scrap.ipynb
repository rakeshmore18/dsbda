```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time

# Setup driver
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))

# Flipkart product review page
url = "https://www.flipkart.com/goboult-mustang-stallion-1-43-amoled-display-functional-bezel-700-nits-brightness-smartwatch/product-reviews/itm58b449eedd78e?pid=ACCFZ95M5JTZQH3F"
driver.get(url)

time.sleep(5)  # wait for page load

# Close login popup if appears
try:
    close_btn = driver.find_element(By.XPATH, "//button[contains(text(),'✕')]")
    close_btn.click()
except:
    pass

time.sleep(3)

# Get all reviews
reviews = driver.find_elements(By.CLASS_NAME, "_27M-vq")

print("Total Reviews:", len(reviews))

for review in reviews:
    try:
        name = review.find_element(By.CLASS_NAME, "_2sc7ZR").text
    except:
        name = "N/A"

    try:
        rating = review.find_element(By.CLASS_NAME, "_3LWZlK").text
    except:
        rating = "N/A"

    try:
        comment = review.find_element(By.CLASS_NAME, "t-ZTKy").text
    except:
        comment = "N/A"

    print("Name:", name)
    print("Rating:", rating)
    print("Comment:", comment)
    print("-" * 50)

driver.quit()
```
