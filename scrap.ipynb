```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
def scrape_demo_books():
    url = "http://books.toscrape.com/catalogue/category/books/science_22/index.html"
    response = requests.get(url)
    
    if response.status_code != 200:
        return []
        
    soup = BeautifulSoup(response.text, 'html.parser')
    dataset = []
    
    book_cards = soup.find_all('article', class_='product_pod')
    
    for card in book_cards:
        title = card.find('h3').find('a')['title']
        price = card.find('p', class_='price_color').text.strip()
        stock = card.find('p', class_='instock availability').text.strip()
        
        rating_tag = card.find('p', class_='star-rating')
        rating_classes = rating_tag['class']
        star_rating = rating_classes[1] if len(rating_classes) > 1 else 'None'
        
        dataset.append({
            'Book Title': title,
            'Price': price,
            'Rating (Stars)': star_rating,
            'Availability': stock
        })
        
    return dataset
```

```python
scraped_data = scrape_demo_books()

if scraped_data:
    df = pd.DataFrame(scraped_data)
    display(df.head(10))
    df.to_csv('demo_books_dataset.csv', index=False)
```
